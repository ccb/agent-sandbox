# Action Castle II — Return to Action Castle

The sequel to Action Castle, ported from Parsely. You return to the castle and work toward the *champion* ending — besting the dragon (by wits or by steel), answering its riddle, and winning the king's favor. This notebook plays the champion walkthrough; the module also ships a `WALKTHROUGH_MARRIAGE` for the alternate ending.

---

This notebook drives the game through its winning walkthrough so you can read the story unfold turn-by-turn, then lets you take the controls. The game logic lives in `text_adventure_games/adventures/action_castle_2.py`; here we just *play* it.

## 1. Load the game

We add the repo root to the path (so `text_adventure_games` imports) and load the game module by file path — the same trick the test suite uses, so it works no matter where Jupyter was started from.

```python
import sys
import importlib.util
from pathlib import Path

# Walk up from the working directory to the repo root (the folder
# that contains `text_adventure_games/`).
root = Path.cwd().resolve()
while not (root / "text_adventure_games").is_dir() and root != root.parent:
    root = root.parent
sys.path.insert(0, str(root))

game_file = root / "text_adventure_games/adventures/action_castle_2.py"
spec = importlib.util.spec_from_file_location("action_castle_2_nb", game_file)
mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(mod)
print("loaded", game_file.name)
```

## 2. Build the world and look around

`build_game()` assembles every room, item, and character and wires up the custom actions. `look` describes where you start.

```python
game = mod.build_game()
game.parser.parse_command("look")
print(f"\nScore: {game.score}/{game.max_score}")
```

## 3. Play the winning walkthrough

The walkthrough (here, `mod.WALKTHROUGH_CHAMPION`) is the list of commands that solves the game. We build a fresh world and feed it the commands one at a time, printing each command (`>>>`) above the game's response so you can follow the solution.

```python
game = mod.build_game()
game.parser.parse_command("look")

for cmd in mod.WALKTHROUGH_CHAMPION:
    print(f"\n>>> {cmd}")
    game.do_command(cmd)
    if game.is_game_over():
        break

print("\n" + "=" * 60)
print(f"WON: {game.is_won()}   GAME OVER: {game.is_game_over()}")
print(f"SCORE: {game.score}/{game.max_score}")
```

## 4. Take the controls

Start a fresh game and drive it yourself. Call `do("...")` with any command — `look`, `go north`, `get lamp`, `inventory`, `examine door`, and so on. Re-run the build cell to start over.

(For a classic blocking prompt-loop instead, call `game.game_loop()`.)

```python
game = mod.build_game()

def do(command):
    """Run one command against the live game."""
    game.do_command(command)

do("look")
```
